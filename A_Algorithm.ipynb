```python
import heapq

def a_star_search(graph, heuristics, start, goal):

    # Priority queue stores tuples of: (f_score, g_score, current_node, path)
    open_set = []
    heapq.heappush(open_set, (heuristics[start], 0, start, [start]))

    # Keeps track of the lowest g_score recorded for each node
    g_scores = {start: 0}

    while open_set:
        f_score, g_score, current, path = heapq.heappop(open_set)

        # Goal check
        if current == goal:
            return path, g_score

        # If we found a worse path to a node we already reached with lower cost, skip
        if g_score > g_scores.get(current, float('inf')):
            continue

        for neighbor, weight in graph.get(current, []):
            tentative_g = g_score + weight

            # If a shorter path to neighbor is found
            if tentative_g < g_scores.get(neighbor, float('inf')):
                g_scores[neighbor] = tentative_g
                f_neighbor = tentative_g + heuristics.get(neighbor, 0)
                heapq.heappush(open_set, (f_neighbor, tentative_g, neighbor, path + [neighbor]))

    return None, float('inf')


if __name__ == "__main__":
    # Graph layout with edge weights
    graph = {
        'A': [('B', 1), ('C', 3)],
        'B': [('D', 1), ('E', 5)],
        'C': [('E', 2)],
        'D': [('F', 3)],
        'E': [('F', 1)],
        'F': []
    }

    # Heuristic estimates (h) to the goal 'F'
    heuristics = {
        'A': 4,
        'B': 2,
        'C': 3,
        'D': 2,
        'E': 1,
        'F': 0
    }

    path, cost = a_star_search(graph, heuristics, start='A', goal='F')

    print(f"Optimal Path: {' -> '.join(path)}")
    print(f"Total Cost:   {cost}")
```
